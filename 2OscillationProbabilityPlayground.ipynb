The detailed derivation of the neutrino oscillation probability is involved, but we will discuss an approximate result that you can use to build intuition and then we will then use an interactive plot of the exact result to further develop intuition over the effects of oscillations.



## 2.1 Neutrino States

<img align="right" style="height:30em;margin-left:5em;" src="nbdata/numassflav.png" />

The key starting point is that the three neutrinos that we know of do not, in fact, have a single set of identities. We have three _flavor_ states: $\nu_\mathrm{e}$, $\nu_\mu$, and $\nu_\tau$, which are the identities used for weak force interactions that also involve a charged lepton: electron, muon, or tau. We _also_ have three _mass_ states: $\nu_1$, $\nu_2$, $\nu_3$, these identities have well defined masses and are used when calculating how neutrinos travel through space. If we have a neutrino in a well defined _flavor_ state, then it must be in a _superposition_ of _mass_ states, and _vice versa_. As an example, if we consider a radioactive beta decay, which produces an electron and a neutrino, then we know that the neutrino must be in a $\nu_\mathrm{e}$ state as it was produced in an interaction involving an electron. As that neutrino travels from it's point of creation, it does so as a probabilistic combination of all three mass states. This might not make a lot of intuitive sense, many of the intermediate steps involved in quantum mechanics don't, but the observable predictions of quantum mechanics are increadibly precise, so even if you just consider it a mathematical tool, it is a very useful one. 

The figure on the right (reproduced from De Gouvêa: https://arxiv.org/pdf/hep-ph/0411274) provides an illustration for this dual nature of neutrinos: The three horizontal bands correspond to the three masses, which we know must differ as the difference between them is what drives oscillation. However, each mass corresponds to some proportion of each of the three flavors, represented by the fraction of each band that is colored. This figure graphically illustrates most of the parameters in neutrino oscillation: the two mass differences and the three different proportions that the three flavors can make up of the three masses. The one parameter that isn't illustrated is actually our parameter of interest, which we will introduce shortly.

## 2.2 Approximate Oscillation Probabilities

For our purposes, we need to know the probability for a muon neutrino of energy, $E$ to interact and produce a charged lepton of flavor, $x$, after having travelled a distance, $L$. For the experiment we are discussing, $\nu_\tau$ are rarely visible and will not be used for the main measurement. So, what we really need to know is just the probabilities for a muon (anti)neutrino to be later detected as a muon (anti)neutrino or as an electron (anti)neutrino, called the muon neutrino survival, $P_{\nu_\mu\left(\bar{\nu}_\mu\right)\rightarrow\nu_\mu\left(\bar{\nu}_\mu\right)}$, and electron neutrino appearance, $P_{\nu_\mu\left(\bar{\nu}_\mu\right)\rightarrow\nu_\mathrm{e}\left(\bar{\nu}_\mathrm{e}\right)}$, probabilities, respectively. 

### 2.2.1 Muon Neutrino Survival

An approxmiate result for the muon neutrino survival probability can be taken as

$$
\def\thot{{{\color{red}\theta_{13}}}}
\def\thtt{{{\color{blue}\theta_{23}}}}
\def\dmto{{{\color{orange}\Delta{}m^2_{31}}}}
\def\dmtwo{{{\color{cyan}\Delta{}m^2_{21}}}}
\def\dmsol{{{\color{cyan}\Delta{}m^2_\mathrm{sol}}}}
\def\dmtt{{{\color{purple}\Delta{}m^2_{32}}}}
\def\dmatm{{{\color{purple}\Delta{}m^2_\mathrm{atm}}}}
\def\dcp{{{\color{green}\delta_\mathrm{CP}}}}
\require{ams}
$$

$$
\begin{align*}
P_{\nu_\mu\left(\bar{\nu}_\mu\right)\rightarrow\nu_\mu\left(\bar{\nu}_\mu\right)}\left(L,E\right) & \simeq 1 - 4\cos^2\thot\sin^2\thtt\left[1 - \cos^2\thot\sin^2\thtt\right]\sin^2\frac{\dmtt{}L}{4E} \\
& + \left[\mathrm{higher\ order\ terms}\right],
\end{align*}
$$

where $\theta_{ij}$ are called the mixing angles and control the proportion of each flavor state within each mass state, $\Delta{}m^{2}_{ij} = m^2_i - m^2_j$ are the mass-squared splittings between the masses of the three mass states, and $\delta_{CP}$ is the Charge-Parity violating phase factor, our parameter of most interest. The equation above is long, complex, and approximate, and you should not expect to build general intuition from just staring at them. The main point to take away is that it is built from trignometric functions of the propagation distance, $L$, over the neutrino Energy, $E$, which gives the probabilities a characteristic _oscillatory_ behavior as a function of $L/E$, as shown below.

### 2.2.2 Electron Neutrino Appearance

Now lets look at the probability for electron (anti)neutrinos to appear in a muon (anti)neutrino beam, 
$$
\def\thot{{{\color{red}\theta_{13}}}}
\def\thtt{{{\color{blue}\theta_{23}}}}
\def\dmto{{{\color{orange}\Delta{}m^2_{31}}}}
\def\dmtwo{{{\color{cyan}\Delta{}m^2_{21}}}}
\def\dmsol{{{\color{cyan}\Delta{}m^2_\mathrm{sol}}}}
\def\dmtt{{{\color{purple}\Delta{}m^2_{32}}}}
\def\dmatm{{{\color{purple}\Delta{}m^2_\mathrm{atm}}}}
\def\dcp{{{\color{green}\delta_\mathrm{CP}}}}
\require{ams}
$$

$$
\begin{align*}
P_{\nu_\mu\left(\bar{\nu}_\mu\right)\rightarrow\nu_\mathrm{e}\left(\bar{\nu}_\mathrm{e}\right)} \left(L,E\right) & \simeq \sin^2\thtt\sin^22\thot\sin^2\frac{\dmto{}L}{4E} \\
& + \sin^22\thtt\sin^22\thot\cos\thot\sin\frac{\dmtwo{}L}{4E}\sin\frac{\dmto{}L}{4E} \\
& \times \left[\cos\frac{\dmtt{}L}{4E}\cos\dcp -\left(+\right) \sin\frac{\dmtt{}L}{4E}\sin\dcp\right] \\
& +  \left[\mathrm{higher\ order\ terms}\right],
\end{align*}
$$

where the only completely new parameter, $\dcp$, is the Charge-Parity violating phase factor, our parameter of most interest. Note that, by definition: $\dmto = \dmtt + \dmtwo$, so really, we can only vary two of these parameters independently, we will choose $\dmtt$ and $\dmtwo$, which can be identified with $\dmatm$ and $\dmsol$ in the earlier figure, respectively.

The very important point here is that the only difference between the predicted oscillations of neutrinos and antineutrinos is the bracketed sign, $-\left(+\right)$ at the end of electron (anti)neutrino appearance probability. This term can generate a difference between the oscillations of neutrinos and antineutrinos if it is nonzero, which is true if and only if $\sin\delta_\mathrm{CP} \neq 0$.

## 2.3 Neutrino Beams

<img align="right" style="width:50em;margin-left:5em;margin-top:5em;" src="nbdata/flux_disco.png" />

To measure neutrino oscillations, we need a source of neutrinos to study. Our focus here is on an experiment that uses an artificial neutrino beam. If you are interested in how we make artifical neutrino beams, the [wikipedia](https://en.wikipedia.org/wiki/Accelerator_neutrino) article is quite good! The important bits for us is that artificial neutrino beams are initially made almost entirely made of muon neutrinos and the beams have some spectral width in energy, we cannot make neutrino beams of a single energy. An example of some real neutrino beam energy distributions for a number of running or planned experiments are shown on the right.

We will approximate the spectral shape of the initially muon-flavor neutrino beam, $\Phi_{\nu_\mu}\left(E\right)$, with a gaussian function, like:
$$\Phi^\mathrm{initial}_{\nu_\mu}\left(E\right) = Ae^{\left[-(E-\alpha)^2/2\beta^2\right]},$$
where A is some arbitrary constant, $\alpha$ controls the peak energy, and $\beta$ controls the width.

The number of neutrinos, of energy $E$, travelling through a detector $L$ km away from the beam source can be found by multiplying by the oscillation probabilities:

$$
\begin{align*}
\Phi^\mathrm{detector}_{\nu_\mu}\left(E, L\right) & = P_{\nu_\mu\rightarrow\nu_\mu}\left(E, L\right)\times\Phi^\mathrm{initial}_{\nu_\mu}\left(E\right)  \\
\Phi^\mathrm{detector}_{\nu_\mathrm{e}}\left(E, L\right) & = P_{\nu_\mu\rightarrow\nu_\mathrm{e}}\left(E, L\right)\times\Phi^\mathrm{initial}_{\nu_\mu}\left(E\right),
\end{align*}
$$

### 2.3.1 Beam Dispersion

We should really make one small, but important, correction to the previous equation: Neutrino beams spread out as they travel. If you shine a torch beam at a wall and move closer and further from the wall, what happens to the size of the spot of light? We are sending these neutrino beams hundreds of kilometers before looking for the effects of oscillation, even with the most perfectly focussed beam, there will be some spread. We often think of laser beams as perfectly focussed beams of light, but [What If?](https://what-if.xkcd.com/13/) you shine a laser beam at the moon? By the time our neutrino beam arrives at our _far_ detector, hundreds of kilometers away from the source, the beam spot is often kilometers wide! We don't need to rigorously derive the reduction in beam intensity due to the spread, but you can assume that it is proportional to $1/L^{2}$, which comes from the surface of a sphere being proportional to it's radius squared. Our small but important correction then is that the number of neutrinos seen at the detector is not just the initial number, but a different number that depends both on $E$ and $L$.

$$
\Phi^\mathrm{initial}_{\nu_\mu}\left(E\right) \rightarrow \Phi^\mathrm{far}_{\nu_\mu}\left(E, L\right)
$$

## 2.4 Cross Sections

<img align="right" style="width:30em;margin-left:5em;" src="nbdata/ubevdisp.png" />

We cannot directly observe the neutrinos that travel through our detector, no matter their flavor. We can only see neutrinos that interact with atoms in our detector and produced secondary charged particles that we can observe with our particle detectors. An example of what the MicroBoone experiment observes when a neutrino hits their detector is shown on the right ([Image Credit: FNAL](https://microboone-exp.fnal.gov/public/approved_plots/plots/8001/run_5235_ev_1915_pl_2.pdf)). The invisible neutrino enters the frame from the left hand side and interacts with an argon nuclei where the three visible lines meet. These lines correspond to charged particles produced in the neutrino-argon interaction and are travelling to the right hand side. 

The fact that we need neutrinos to interact in our detector for us to see them adds a final term to our equations; the _cross section_. As particle physicists, when we talk about the _cross section_ for an interaction, we just mean the probability for that interaction to occur. The name, cross section, comes from comparing particle interactions to the probability of two balls to collide if thrown towards each other. The larger the cross section of the balls, the more likely they are to collide. The cross section, which has units of area, is defined as

$$
\sigma = \frac{R}{\Phi},
$$

where $R$ is the number of interactions, and $\Phi$ is the number of neutrinos. It looks just like a simple probability, the number of times something happened (number of interaction) divided by the number of chances that it had to happen (number of neutrinos). A small but important extra detail is that, $\Phi$ is really a _flux_ of neutrinos, so it is the number of neutrinos per unit area, which gives $\sigma$ the units of area. We should expect this interaction probability to depend on the energy, $\sigma\left(E\right)$. We can then rearrange this equation to calculate the average number of expected interactions if we know the cross section and the number of neutrinos:

$$
\begin{align*}
R^\mathrm{detector}_{\nu_\mu}\left(E, L\right) & = P_{\nu_\mu\rightarrow\nu_\mu}\left(E, L\right)\times\Phi^\mathrm{far}_{\nu_\mu}\left(E, L\right) \times \sigma \\
R^\mathrm{detector}_{\nu_\mathrm{e}}\left(E, L\right) & = P_{\nu_\mu\rightarrow\nu_\mathrm{e}}\left(E, L\right)\times\Phi^\mathrm{far}_{\nu_\mu}\left(E, L\right) \times \sigma,
\end{align*}
$$

where we are treating the cross section for muon neutrinos and electron neutrinos to interact as the same. This is a fairly good approximation. However, the cross section for neutrinos and antineutrinos differ quite signficiantly, as can be seen in the figure below which contains the results of a theoretical calculation for the cross sections for neutrinos, $\sigma_\nu$, and antineutrinos, $\sigma_\bar{\nu}$, to interact with an Argon nuclei and produce a charged lepton, _e.g._ 
$$\nu_\mu + {}^{40}\mathrm{Ar} \rightarrow \mu^- + {}^{39}\mathrm{Ar} + \mathrm{p}^+$$ or 
$$\bar{\nu}_\mathrm{e} + {}^{40}\mathrm{Ar} \rightarrow \mathrm{e}^+ + {}^{39}\mathrm{Cl} + \mathrm{n},$$
where we are using an overbar, like $\bar{\nu}$, to denote an antiparticle, and $\mathrm{p}$ and $\mathrm{n}$ refer to protons and neutrons, respectively.
